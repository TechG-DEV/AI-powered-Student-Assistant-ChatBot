```python
import json 
import os
FileName = 'full_dataset/full_dataset.json'
FileName2 = 'context_dataset/context_dataset.json'

def contextcr(new_entries):
    
    if os.path.exists('context_dataset/context_dataset.json'): 
        with open(FileName2, 'r') as json_file:
            existing_data = json.load(json_file)

    else:
        existing_data = []

    existing_data.extend(new_entries)

    with open(FileName2, 'w') as json_file:
        json.dump(existing_data, json_file, indent=4)




def jsoncr(new_entries):
    
    if os.path.exists('full_dataset/full_dataset.json'): 
        with open(FileName, 'r') as json_file:
            existing_data = json.load(json_file)

    else:
        existing_data = []

    existing_data.extend(new_entries)

    with open(FileName, 'w') as json_file:
        json.dump(existing_data, json_file, indent=4)
```

```python
import tkinter as tk

root = tk.Tk(className = 'Dataset Creation')
root.geometry("1000x1000")

def insbutt():
    if text3.get("1.0", "end-1c") == x:
        cntx = None
    else:
        cntx = text3.get("1.0", "end-1c")

    entries = [{
        'question': text1.get("1.0", "end-1c"),
        'context': cntx,
        'answer': text2.get("1.0", "end-1c")
    }]
    jsoncr(entries)
    
    for key,value in entries[0].items():
        if key == 'context':
            cntxlist = [{
                key : value
            }]
    contextcr(cntxlist)
    



button = tk.Button(root, text = "Inster To Dataset", command = insbutt)
button.pack(pady= 200)

label1 = tk.Label(root, text= "Question:")
label1.pack()
text1 = tk.Text(root, height= 5, width= 52)
text1.pack()
text1.insert( tk.END,'place question here')

label2 = tk.Label(root, text= "Context:")
label2.pack()
text2 = tk.Text(root, height= 5, width= "52")
text2.pack()


label3 = tk.Label(root, text= "Answer:")
label3.pack()
text3 = tk.Text(root, height= 5, width= 52)
text3.pack()
text3.insert( tk.END,'place answer here')

x = text3.get('1.0','end-1c')


root.mainloop()
```
